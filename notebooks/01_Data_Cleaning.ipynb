# Superstore Sales Analysis - Data Cleaning

**Goal:** load the raw Superstore dataset, check its quality, remove problems, and save a clean copy for the EDA, SQL, and Power BI steps.

**Steps:**
1. Load and inspect the raw data
2. Remove duplicate rows and trim whitespace
3. Validate the cleaned data
4. Save the cleaned data as a formatted Excel file and as a CSV file

## 1. Load and Inspect the Raw Data

```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv("../data/Raw/SampleSuperstore.csv")

# Display first 5 rows
print("First 5 rows:")
display(df.head())

# Dataset shape
print("\nDataset Shape:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

**Findings:** the raw file has **9,994 rows and 13 columns**. There are **no missing values**, and the data types are correct (text columns, numeric Sales, Quantity, Discount and Profit). The inspection found **17 duplicate rows**, which are removed in the next step.

## 2. Clean the Data

```python
# ============================================
# SUPERSTORE DATA CLEANING
# ============================================

# Create a copy of the raw data
cleaned_df = df.copy()

# Count duplicate rows
duplicate_count = cleaned_df.duplicated().sum()
print(f"Duplicate rows found: {duplicate_count}")

# Remove exact duplicate rows
cleaned_df = cleaned_df.drop_duplicates()

# Check missing values
print("\nMissing values after duplicate removal:")
print(cleaned_df.isnull().sum())

# Remove unnecessary whitespace from text columns
text_columns = cleaned_df.select_dtypes(include="object").columns

for column in text_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()

# Check data types
print("\nData types:")
print(cleaned_df.dtypes)

# Check cleaned dataset shape
print("\nCleaned dataset shape:")
print(cleaned_df.shape)

# Check duplicates again
print("\nDuplicate rows after cleaning:")
print(cleaned_df.duplicated().sum())
```

**Result:** the 17 exact duplicate rows were removed and extra spaces were trimmed from text columns. The cleaned dataset has **9,977 rows** and still has 0 missing values.

## 3. Validate the Cleaned Data

```python
# ============================================
# VALIDATE CLEANED DATA
# ============================================

print("First 5 rows of cleaned data:")
display(cleaned_df.head())

print("\nLast 5 rows of cleaned data:")
display(cleaned_df.tail())

print("\nCleaned dataset shape:")
print(cleaned_df.shape)

print("\nMissing values:")
print(cleaned_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(cleaned_df.duplicated().sum())

print("\nBasic statistics:")
display(cleaned_df.describe())
```

**Validation:** shape is **(9,977, 13)**, there are **0 missing values** and **0 duplicate rows**. The statistics look reasonable (Discount between 0 and 0.80, Quantity between 1 and 14). Profit has large negative and positive values, which are real sales results and are kept.

## 4. Save the Cleaned Data (Excel)

```python
# ============================================
# SAVE CLEANED DATASET AS FORMATTED EXCEL
# ============================================

import os
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.worksheet.table import Table, TableStyleInfo

# Create Cleaned folder if it doesn't exist
cleaned_folder = "../data/Cleaned"
os.makedirs(cleaned_folder, exist_ok=True)

# Excel file path
cleaned_file_path = "../data/Cleaned/Superstore_Cleaned.xlsx"

# Save cleaned DataFrame to Excel
cleaned_df.to_excel(
    cleaned_file_path,
    index=False,
    sheet_name="Cleaned_Data"
)

# Open the Excel file for formatting
wb = load_workbook(cleaned_file_path)
ws = wb["Cleaned_Data"]


# ============================================
# 1. FORMAT HEADER
# ============================================

for cell in ws[1]:
    cell.font = Font(
        bold=True,
        color="FFFFFF"
    )

    cell.fill = PatternFill(
        fill_type="solid",
        fgColor="1F4E78"
    )

    cell.alignment = Alignment(
        horizontal="center",
        vertical="center"
    )


# ============================================
# 2. SET HEADER ROW HEIGHT
# ============================================

ws.row_dimensions[1].height = 25


# ============================================
# 3. ADJUST COLUMN WIDTH AUTOMATICALLY
# ============================================

for column_cells in ws.columns:

    max_length = 0
    column_letter = column_cells[0].column_letter

    for cell in column_cells:
        if cell.value is not None:
            max_length = max(
                max_length,
                len(str(cell.value))
            )

    # Add extra space
    adjusted_width = max_length + 2

    # Prevent extremely wide columns
    adjusted_width = min(adjusted_width, 30)

    ws.column_dimensions[column_letter].width = adjusted_width


# ============================================
# 4. SET ROW HEIGHT
# ============================================

for row in range(2, ws.max_row + 1):
    ws.row_dimensions[row].height = 20


# ============================================
# 5. FORMAT SALES
# ============================================

sales_column = cleaned_df.columns.get_loc("Sales") + 1
sales_letter = ws.cell(
    row=1,
    column=sales_column
).column_letter

for cell in ws[sales_letter][1:]:
    cell.number_format = '#,##0.00'


# ============================================
# 6. FORMAT PROFIT
# ============================================

profit_column = cleaned_df.columns.get_loc("Profit") + 1
profit_letter = ws.cell(
    row=1,
    column=profit_column
).column_letter

for cell in ws[profit_letter][1:]:
    cell.number_format = '#,##0.00'


# ============================================
# 7. FORMAT DISCOUNT AS PERCENTAGE
# ============================================

discount_column = cleaned_df.columns.get_loc("Discount") + 1
discount_letter = ws.cell(
    row=1,
    column=discount_column
).column_letter

for cell in ws[discount_letter][1:]:
    cell.number_format = '0%'


# ============================================
# 8. CENTER NUMERIC COLUMNS
# ============================================

for row in ws.iter_rows(min_row=2):

    # Postal Code
    row[5].alignment = Alignment(horizontal="center")

    # Quantity
    row[10].alignment = Alignment(horizontal="center")

    # Discount
    row[11].alignment = Alignment(horizontal="center")


# ============================================
# 9. FREEZE HEADER ROW
# ============================================

ws.freeze_panes = "A2"


# ============================================
# 10. ADD FILTERABLE EXCEL TABLE
# ============================================

last_row = ws.max_row
last_column = ws.max_column

table_reference = (
    f"A1:{ws.cell(row=last_row, column=last_column).coordinate}"
)

table = Table(
    displayName="SuperstoreCleanedData",
    ref=table_reference
)

table_style = TableStyleInfo(
    name="TableStyleMedium2",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False
)

table.tableStyleInfo = table_style

ws.add_table(table)


# ============================================
# 11. SAVE FINAL FILE
# ============================================

wb.save(cleaned_file_path)


# ============================================
# 12. CONFIRMATION
# ============================================

print("============================================")
print("CLEANED EXCEL FILE SAVED SUCCESSFULLY")
print("============================================")
print(f"File location: {cleaned_file_path}")
print(f"Rows saved: {len(cleaned_df)}")
print(f"Columns saved: {len(cleaned_df.columns)}")
print(f"Duplicates: {cleaned_df.duplicated().sum()}")
print(f"Missing values: {cleaned_df.isnull().sum().sum()}")
```

## 5. Save the Cleaned Data (CSV)

The CSV copy is used to import the data into PostgreSQL for the SQL analysis (PostgreSQL cannot read `.xlsx` files directly).

```python
# ============================================
# SAVE CLEANED DATASET AS CSV (FOR POSTGRESQL)
# ============================================

cleaned_csv_path = "../data/Cleaned/Superstore_Cleaned.csv"

cleaned_df.to_csv(cleaned_csv_path, index=False)

print("============================================")
print("CLEANED CSV FILE SAVED SUCCESSFULLY")
print("============================================")
print(f"File location: {cleaned_csv_path}")
print(f"Rows saved: {len(cleaned_df)}")
print(f"Columns saved: {len(cleaned_df.columns)}")
```
